### Few Begginer Commands
 1) To check the Version of GIT - git --version.
 2) To Check the Files & Folders - ls
 3) To delete all the commands - clear
 4) To check the Current Working Directory - pwd

### Git Configuration
There are two major command's that connect your Git with GITHUB.<br>
user_name ----> git config --global user.name "mr-amit-yadav"<br>
user_email ----> git config --global user.email "user@gmail.com"

### Clone 
To clone a repository on local machine.<br>
git clone <-Some Link-><br>
git clone "https://mr-amit-yadav/Predictive-Analytics"


### Changing the Directory on Local Machine.
There are  4 to 5 major commads for this process.<br>
To check the current directory - pwd.<br>
If you are inside a folder, then to come out of folder - cd ..<br>
Next step is to use - cd ~<br>
To get into a particular folder - cd folder_name.<br>
Note: To understand the flow of the process keep on checking by using pwd command in which directory you really are into.<br>

Something Extra:
1) ls --To get the list of files.<br>
2) ls-a -- To see the hidden files.<br>
3)ls -Force --Alternate option in some of the system where ls-a command doesn't work.<br>
